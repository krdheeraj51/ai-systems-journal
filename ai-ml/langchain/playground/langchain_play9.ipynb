```python
from langchain_openai import AzureChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import tool
from langchain_core.output_parsers import StrOutputParser
import os 
from dotenv import load_dotenv
load_dotenv()
```

```python
## Create tools 
@tool
def add(num1: int, num2: int) -> int:
    """Adds two numbers together."""
    return num1 + num2


@tool
def multiply(num1: int, num2: int) -> int:
    """Multiplies two numbers together."""
    return num1 * num2

@tool
def square(num: int) -> int:
    """Squares a number."""
    return num * num
```

```python
# --- IGNORE ---
azure_deployment = os.getenv("AZURE_OPENAI_DEPLOYMENT_NAME")
azure_api_version = os.getenv("AZURE_OPENAI_API_VERSION")

model = AzureChatOpenAI(azure_deployment=azure_deployment, api_version=azure_api_version)
prompt = ChatPromptTemplate.from_messages([
    ("system", """
You are an intelligent assistant.

You can:
- Break problems into steps
- Use tools multiple times
- Think step-by-step before answering
"""),
    ("user", "{input}")
])

```

```python
model_with_tool = model.bind_tools([add, multiply, square])
```

```python
# from langchain.agents.agent import AgentExecutor
# agent_chain = prompt | model_with_tool 



from langchain_classic.agents import create_tool_calling_agent
agent = create_tool_calling_agent(llm=model, tools=[add, multiply, square], prompt=prompt)

response = agent.invoke({
    "input": "Add 5 and 3, then multiply the result by 10, then square it"
})

print(response)
```
