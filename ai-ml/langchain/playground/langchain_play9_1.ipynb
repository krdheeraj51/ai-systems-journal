```python
#import all required packages 
from langchain_openai import AzureChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import tool
from langchain_core.output_parsers import StrOutputParser
import os 
from dotenv import load_dotenv
load_dotenv()
```

```python
#define tools to be used in the agent
@tool 
def add(num1: float, num2: float) -> float:
    """Add two numbers."""
    return num1 + num2

@tool
def multiply(num1: float, num2: float) -> float:
    """Multiply two numbers."""
    return num1 * num2

@tool
def square(num: float) -> float:
    """Square a number."""
    return num * num
```

```python
import langchain
import langchain_openai
print(langchain.__version__)
```

```python
#load the Azure OpenAI API key and endpoint from environment variables
azure_deployment = os.getenv("AZURE_OPENAI_DEPLOYMENT_NAME")
azure_api_version = os.getenv("AZURE_OPENAI_API_VERSION")

model = AzureChatOpenAI(azure_deployment=azure_deployment, api_version=azure_api_version)

prompt_template = ChatPromptTemplate.from_messages([
    ("system", """
    You are an intelligent assistant.

    You can:
    - Break problems into steps
    - Use tools multiple times
    - Think step-by-step before answering
    """),
    ("user", "{input}")
])
```

```python
model_with_tools = model.bind_tools([add, multiply, square])
chain = prompt_template | model_with_tools
response = chain.invoke({
    "input": "Add 5 and 3, then multiply the result by 10, then square it"
})

print(response)
```
